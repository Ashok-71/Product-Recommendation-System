# 📊 EDA — Product Recommendation System
Exploratory analysis of the product review dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys, os
sys.path.insert(0, '..')
from src.data_preprocessing import load_or_generate, get_product_catalogue

df = load_or_generate('data/raw/products.csv', 'data/processed/cleaned_data.csv')
catalogue = get_product_catalogue(df)
print(df.shape)
df.head()
```

```python
# Rating distribution
df['rating'].value_counts().sort_index().plot(kind='bar', color='steelblue')
plt.title('Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Top categories
df.groupby('category')['product_id'].count().sort_values().plot(kind='barh')
plt.title('Reviews per Category')
plt.tight_layout()
plt.show()
```

```python
# Price distribution
catalogue['price'].hist(bins=40, color='darkorange', edgecolor='black')
plt.title('Product Price Distribution')
plt.xlabel('Price (INR)')
plt.tight_layout()
plt.show()
```
